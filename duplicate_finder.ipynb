```python
import filecmp as fcp
import glob
import pandas as pd
```

```python
img_dir_path = "/projekt1/ag_maahn/data_obs/lim/allsky1/images/2026/**/*.png"

filepath_collector = glob.glob(img_dir_path, recursive=True)
filepath_collector.sort()

print("Found {0} files".format(len(filepath_collector)))

duplicate_collector = []

for i in range(1, len(filepath_collector)):

    if fcp.cmp(filepath_collector[i], filepath_collector[i-1], shallow=True):

        duplicate_collector.append(filepath_collector[i])

    print(" Running - {0} image pairs compared - {1} duplicates found".format(i, len(duplicate_collector)), end="\r")

print("")
print("Done")
```

```python
df = pd.DataFrame({"Duplicate":duplicate_collector})

df.to_csv("duplicates_out.csv")
```
